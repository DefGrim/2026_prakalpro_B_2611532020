## 2.1 Landasan Teori

**1. Variabel**

Variabel adalah wadah atau lokasi di dalam memori komputer yang dialokasikan untuk menyimpan suatu nilai data. Nilai yang ditampung oleh variabel bersifat dinamis, yang berarti nilai tersebut dapat diubah, diperbarui, atau dipanggil kembali sewaktu-waktu selama siklus eksekusi program berlangsung. Penggunaan variabel sangat penting agar kode program lebih terstruktur dan efisien karena data tidak perlu ditulis ulang secara manual setiap kali dibutuhkan dalam operasi komputasi.

**2. Tipe Data**

Tipe data merupakan klasifikasi atau pengelompokan yang menentukan jenis nilai dari suatu variabel, sehingga interpreter Python dapat memahami bagaimana data tersebut harus diproses. Python menerapkan sistem dynamic typing, di mana tipe data dikenali dan ditentukan secara otomatis oleh sistem berdasarkan nilai yang diberikan saat inisialisasi variabel tanpa perlu deklarasi tipe secara eksplisit.

**3. Jenis-Jenis Tipe Data Dasar**

Python memiliki berbagai macam tipe data esensial yang sering digunakan dalam pemrograman, antara lain:

Integer (int): Digunakan untuk merepresentasikan bilangan bulat, baik positif, negatif, maupun nol (contoh: 100, -5).

Float (float): Digunakan untuk bilangan pecahan atau desimal yang menggunakan tanda titik sebagai pemisah (contoh: 10.2345, 3.14).

String (str): Digunakan untuk menyimpan data teks atau kumpulan karakter yang diapit menggunakan tanda petik tunggal maupun ganda (contoh: "Python", 'L').

Boolean (bool): Tipe data logika yang hanya merepresentasikan dua kondisi kebenaran, yaitu True (benar) atau False (salah).

Complex (complex): Digunakan untuk bilangan kompleks yang terdiri dari bagian riil dan imajiner (contoh: 100+3j).

**4. Konstanta dan Fungsi Input**

Selain variabel dan tipe data dasar, pengembangan program sering kali melibatkan penggunaan konstanta (seperti modul Final pada Python) untuk menyimpan nilai tetap yang tidak boleh diubah sepanjang program berjalan. Selain itu, fungsi input() digunakan untuk menerima data atau nilai secara interaktif langsung dari pengguna saat program dieksekusi.

## 2.2 Kode Program

### 2.2.1 Program "Boolean_2611532020"

```python
# Buat file dengan nama Boolean_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: nilai_1234
# Deklarasi variabel dengan tipe data Boolean
is_lulus = True
is_cumlaude = True

# Menggunakan Boolean
nilai_2020 = 85
batas_lulus_2020 = 75

# Menentukan nilai Booelean dari kondisi
status_kelulusan_2020 = nilai_2020 >= batas_lulus_2020 # Hasilnya akan True

print("=== Check Kelulusan ===")
print("Nilai:", nilai_2020)
print("Apakah lulus?", status_kelulusan_2020)
if is_lulus and is_cumlaude:
    print("Selamat! Anda lulus dengan predikat Cum Laude!")
```

**Penjelasan :**
1. Baris 1 sampai 3 berfungsi untuk membuat variabel Boolean is_lulus dengan nilai True dan is_cumlaude dengan nilai True.
2. Baris 5 sampai 7 berfungsi untuk menyimpan nilai mahasiswa dan batas nilai kelulusan ke dalam variabel.
3. Baris 9 sampai 10 berfungsi untuk menentukan nilai Boolean dari kondisi perbandingan nilai dengan batas lulus dan menyimpannya ke dalam variabel status_kelulusan.
4. Baris 12 sampai 18 berfungsi untuk menampilkan judul, nilai mahasiswa, hasil status kelulusan ke layar, serta menjalankan pengecekan percabangan untuk mencetak pesan ucapan selamat lulus predikat Cum Laude.

### 2.2.2 Program "Input_2611532020"

```python
# Buat file dengan nama Input_NIM.py
# Buat program untuk menghitung luas persegi panjang
# Nama variabel ditambah 4 digit nim terakhir contoh: panjang_2020
# Program ini menggunakan fungsi input() untuk meminta penguna
# memasukkan nilai panjang dan lebar persegi panjang.
# Nilai yang dimasukkan akan dikorversi menjadi tipe data integer
panjang_2020 = input('Masukkan nilai panjang: ')
lebar_2020 = input('Masukkan nilai lebar: ')
luas_2020 = int(panjang_2020) * int(lebar_2020)
print("Luas =", luas_2020)
```

**Penjelasan :**
1. Baris 1 sampai 2 berfungsi untuk meminta pengguna memasukkan nilai panjang dan lebar menggunakan fungsi input().
2. Baris 3 berfungsi untuk mengonversi nilai input panjang dan lebar yang bertipe string menjadi integer, lalu mengalikannya untuk menghitung luas persegi panjang.
3. Baris 4 berfungsi untuk menampilkan hasil perhitungan luas persegi panjang ke layar.

### 2.2.3 Program "Konstanta_2611532020"

```python
# Buat file dengan nama Konstanta_NIM.py
# Program in menggunakan konstanta untuk menghitung luas lingkaran
# Nama variabel ditambah 4 digit nim terakhir contoh: jari_2020

from typing import Final
PI: Final = 3.14
print("pi: %f" % (PI))
jari_2020 = float(input("Masukkan nilai jari-jari: "))
luas_2020 = PI * jari_2020 * jari_2020
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_2020, luas_2020))
```

**Penjelasan :**
1. Baris 1 sampai 3 berfungsi untuk mengimpor modul Final dan mendeklarasikan konstanta PI dengan nilai tetap 3.14, lalu mencetaknya ke layar.
2. Baris 4 berfungsi untuk meminta pengguna memasukkan nilai jari-jari lingkaran melalui fungsi input() dan mengonversinya ke tipe data float.
3. Baris 5 berfungsi untuk menghitung luas lingkaran menggunakan rumus perkalian konstanta PI dengan kuadrat nilai jari-jari.
4. Baris 6 berfungsi untuk menampilkan hasil akhir perhitungan luas lingkaran dengan format dua angka di belakang koma ke layar.

### 2.2.4 Program "Numerik_2611532020"

```python
# Buat file dengan nama Numerik_Nim.py
# nama variabel ditambah 4 digit terakhir NIM contoh: a_5500
#create a variable with numerix type value.
a_2020=100
print("The type of variable having value", a_2020, " is ", type(a_2020))

#create a variable with float value
b_2020=10.2345
print("The type of variable having value", b_2020, " is ", type(b_2020))

#create a variable with complex value.
c_2020=100+3j
print("The type of variable having value", c_2020, " is ", type(c_2020))
```

**Penjelasan :**
1. Baris 1 sampai 3 berfungsi untuk membuat variabel bertipe data numerik integer (a_2020) dan mencetak nilai serta tipe datanya ke layar.
2. Baris 5 sampai 7 berfungsi untuk membuat variabel bertipe data pecahan atau float (b_2020) dan mencetak nilai serta tipe datanya ke layar.
3. Baris 9 sampai 11 berfungsi untuk membuat variabel bertipe data kompleks (c_2020) yang memiliki bagian riil dan imajiner, lalu mencetak nilai serta tipe datanya ke layar.
