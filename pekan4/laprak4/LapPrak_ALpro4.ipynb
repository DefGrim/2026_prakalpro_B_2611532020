## 2.1 Landasan Teori

**2.1.1 Konsep Dasar Struktur Percabangan** 

Secara umum, alur eksekusi program komputer berjalan secara sekuensial atau berurutan dari baris kode pertama hingga terakhir. Namun, dalam menghadapi permasalahan nyata, sebuah program sering kali harus mampu mengambil keputusan berdasarkan kondisi tertentu. Struktur percabangan (branching atau conditional statements) adalah mekanisme dalam pemrograman yang memungkinkan blok kode tertentu dieksekusi hanya jika suatu kondisi bernilai benar (True), dan dilewati atau dialihkan ke jalur lain jika bernilai salah (False). Dalam bahasa pemrograman Python, penentuan blok kode di dalam struktur percabangan sangat bergantung pada aturan pengetikan indentasi (spasi atau tab) yang konsisten.

**2.1.2 Pernyataan if (Percabangan Tunggal)**

Pernyataan if merupakan bentuk struktur percabangan paling sederhana yang digunakan untuk menguji sebuah kondisi tunggal. Jika ekspresi logika yang dievaluasi menghasilkan nilai True, maka blok pernyataan di dalam cakupan if akan dieksekusi oleh program. Sebaliknya, jika kondisi bernilai False, blok kode tersebut akan diabaikan dan program langsung melanjutkan ke baris kode berikutnya di luar blok percabangan.

**2.1.3 Pernyataan if-else (Percabangan Dua Arah)**

Pernyataan if-else digunakan ketika sebuah program memerlukan dua jalur alternatif eksekusi berdasarkan satu kondisi. Jika ekspresi logika pada if bernilai True, maka blok kode pertama yang akan dijalankan. Namun, apabila kondisi tersebut bernilai False, program secara otomatis akan beralih untuk mengeksekusi blok kode alternatif yang berada di dalam bagian else. Struktur ini memastikan bahwa salah satu dari dua jalur kode pasti akan dieksekusi.

**2.1.4 Pernyataan if-elif-else (Percabangan Bertingkat / Multi-Kondisi)**

Pernyataan if-elif-else digunakan ketika terdapat lebih dari dua alternatif kondisi yang harus diuji secara berurutan. Elif merupakan singkatan dari else-if. Program akan melakukan pengecekan dari atas ke bawah; kondisi pertama yang bernilai True akan dieksekusi, dan sisa kondisi di bawahnya akan diabaikan. Jika seluruh kondisi di atas bernilai False, maka blok else di bagian akhir akan dieksekusi sebagai pilihan terakhir (default).

**2.1.5 Percabangan Multi-If**

Percabangan multi-if melibatkan beberapa pernyataan if terpisah yang ditulis secara berurutan dalam satu level blok kode. Berbeda dengan if-elif-else yang bersifat eksklusif (hanya satu blok yang pasti dijalankan), struktur multi-if mengevaluasi setiap kondisi secara independen. Artinya, meskipun kondisi pertama bernilai True dan blok kodenya dieksekusi, program akan tetap memeriksa kondisi-kondisi if berikutnya satu per satu, sehingga memungkinkan lebih dari satu blok kode untuk dieksekusi secara bersamaan jika kondisinya terpenuhi.

**2.1.6 Operator Logika (and, or, not)**

Dalam menyusun kondisi percabangan yang kompleks, sering kali diperlukan penggabungan lebih dari satu ekspresi boolean. Bahasa pemrograman Python menyediakan tiga operator logika utama:

1. `and` (Konjungsi): Menghasilkan nilai True hanya jika semua kondisi yang dihubungkan bernilai True.

2. `or` (Disjungsi): Menghasilkan nilai True jika salah satu atau kedua kondisi yang dihubungkan bernilai True.

3. `not` (Negasi): Digunakan untuk membalikkan nilai kebenaran suatu kondisi; mengubah True menjadi False dan sebaliknya.

**2.1.7 Pernyataan match-case**

Pernyataan match-case merupakan fitur percabangan modern yang diperkenalkan mulai Python versi 3.10. Fitur ini berfungsi sebagai alternatif yang lebih bersih dan terstruktur untuk menggantikan struktur if-elif-else yang terlalu panjang saat menangani pencocokan nilai diskrit (pattern matching). Dengan menggunakan match untuk mengevaluasi sebuah variabel dan beberapa case untuk menentukan nilai kecocokannya, kode program menjadi jauh lebih mudah dibaca, elegan, dan terhindar dari kerumitan penulisan kondisi yang berulang.

### 2.2.5 Program "Multi_if2_2611532020"

```python
# Buat file dengan nama multi_if2_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: total_belanja_1234
# Program ini menggunakan fungsi input ()
# Program Menghitung Diskon Belanja

# Input dari user
total_belanja_2020 = float(input("Masukkan total belanja (Rp): "))

# Input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_2020 = input("Apakah Anda member? (y/t): ").strip().lower()
is_member_2020 = input_member_2020 in ["y", "ya"]

# Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_2020 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_2020 = input_promo_2020 in ["y", "ya"]

total_diskon_persen_2020 = 0

# Multi-IF terpisah: Setiap kondisi diperiksa secara independen
# Diskon bisa ditumbuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_2020 > 1000000:
    total_diskon_persen_2020 += 10 # Diskon belanja besar

if is_member_2020:
    total_diskon_persen_2020 += 5 # Diskon member
    
if kode_promo_valid_2020:
    total_diskon_persen_2020 += 15 # Diskon voucher
    
# Menghitung nominal diskon dan total bayar
nominal_diskon_2020 = total_belanja_2020 * (total_diskon_persen_2020 / 100)
total_bayar_2020 = total_belanja_2020 - nominal_diskon_2020

# Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon  : {total_diskon_persen_2020}% (Rp {nominal_diskon_2020:,.0f})")
print(f"Total Bayar  : Rp {total_bayar_2020:,.0f}")

print(f"Total diskon yang Anda dapatkan: {total_diskon_persen_2020}%")
# Output: Total diskon yang Anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
```

```python
# Buat file dengan nama match_case1_nim.py
# Buat program untuk match case
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input ()
# Program konversi angka menjadi nama bulan

bulan_2020 = int(input("Masukkan angka bulan (1 - 12): "))

match bulan_2020:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```

```python
print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")

nama_pengunjung_2020 = input("Masukkan Nama Pengunjung        : ").strip()
umur_2020 = int(input("Input umur anda                 : "))

print(f"Nama Pengunjung : {nama_pengunjung_2020}")
print(f"Umur            : {umur_2020} tahun")

sim_2020 = input("Apakah Anda Sudah Punya SIM C (y/t): ").strip().lower()
if sim_2020:
    sim_2020 = sim_2020[0]

print()
print("Pilihan Paket Wahana (1-5):")
print("  1. Safari Rimba         (Rp 50,000)")
print("  2. Arung Jeram          (Rp 75,000)")
print("  3. Motor ATV Ekstrim    (Rp 120,000)")
print("  4. Roller Coaster Kilat (Rp 100,000)")
print("  5. All-Access VIP       (Rp 220,000)")

paket_pilihan_2020 = int(input("Masukkan nomor paket (1-5)      : "))
jumlah_tiket_2020 = int(input("Masukkan jumlah tiket           : "))

if jumlah_tiket_2020 <= 0:
    print("Kuota tiket tidak valid.")
    raise SystemExit

is_member_2020 = input("Apakah Anda member? (y/t)       : ").strip().lower()
if is_member_2020:
    is_member_2020 = is_member_2020[0]

kode_promo_valid_2020 = input("Apakah kode promo valid? (y/t)  : ").strip().lower()
if kode_promo_valid_2020:
    kode_promo_valid_2020 = kode_promo_valid_2020[0]

match paket_pilihan_2020:
    case 1:
        nama_paket_2020 = "Wahana Safari Rimba"
        harga_satuan_2020 = 50000
    case 2:
        nama_paket_2020 = "Wahana Arung Jeram"
        harga_satuan_2020 = 75000
    case 3:
        nama_paket_2020 = "Wahana Motor ATV Ekstrim"
        harga_satuan_2020 = 120000
    case 4:
        nama_paket_2020 = "Wahana Roller Coaster Kilat"
        harga_satuan_2020 = 100000
    case 5:
        nama_paket_2020 = "Wahana All-Access VIP"
        harga_satuan_2020 = 220000
    case _:
        print("Paket wahana tidak valid!")
        raise SystemExit

print()
print("--- KELAYAKAN PENGENDARA WAHANA ---")

if paket_pilihan_2020 == 3:
    if umur_2020 >= 17 and sim_2020 == 'y':
        print("Status Akses: Anda sudah dewasa dan boleh mengendarai ATV sendiri.")
    elif umur_2020 >= 17 and sim_2020 != 'y':
        print("Status Akses: Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur).")
    elif umur_2020 < 17 and sim_2020 == 'y':
        print("Status Akses: Identitas tidak valid: Belum cukup umur memiliki SIM.")
    else:
        print("Status Akses: Anda belum cukup umur dan tidak boleh bawa motor ATV.")
else:
    if umur_2020 >= 10:
        print("Status Akses: Anda memenuhi syarat usia untuk wahana ini.")
    else:
        print("Status Akses: Anda belum cukup umur untuk wahana ini.")

subtotal_2020 = harga_satuan_2020 * jumlah_tiket_2020
total_diskon_persen_2020 = 0

if subtotal_2020 >= 200000:
    total_diskon_persen_2020 += 10
if is_member_2020 in ['y', 'ya']:
    total_diskon_persen_2020 += 5
if kode_promo_valid_2020 in ['y', 'ya']:
    total_diskon_persen_2020 += 15
if jumlah_tiket_2020 >= 5:
    total_diskon_persen_2020 += 5

nominal_diskon_2020 = subtotal_2020 * (total_diskon_persen_2020 / 100)
total_bayar_2020 = subtotal_2020 - nominal_diskon_2020

print()
print("--- Rincian Pembayaran ---")
print(f"Subtotal Belanja : Rp {subtotal_2020:,.0f}")
print(f"Total Diskon     : {total_diskon_persen_2020}% (Rp {nominal_diskon_2020:,.0f})")
print(f"Total Bayar      : Rp {total_bayar_2020:,.0f}")

print("Catatan Layanan  : ", end="")
if total_bayar_2020 > 300000:
    print("Selamat! Anda berhak mendapatkan Souvenir Gratis.")
else:
    print("Terima kasih telah berkunjung.")

print("Program Selesai")
```
