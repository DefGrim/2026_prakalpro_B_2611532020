## 2.1 Landasan Teori

**2.1.1 Konsep Dasar Perulangan `for`**

Perulangan (*looping*) adalah struktur kontrol dalam pemrograman yang digunakan untuk mengeksekusi blok kode secara berulang-ulang selama kondisi tertentu terpenuhi atau sebanyak jumlah iterasi yang ditentukan.

Dalam bahasa pemrograman Python, pernyataan `for` dirancang khusus untuk melakukan iterasi terhadap objek yang dapat diulang (*iterable*), seperti *list*, *tuple*, *string*, atau rentang angka. Sintaks umum perulangan `for` adalah sebagai berikut:

Python

```
for variabel in iterable:
    # blok kode yang akan dieksekusi
```

Setiap elemen di dalam *iterable* akan diambil secara berurutan dan dimasukkan ke dalam variabel iterasi, kemudian blok kode di bawahnya akan dijalankan hingga elemen terakhir tercapai.

**2.1.2 Fungsi `range()`**

Fungsi `range()` adalah fungsi bawaan Python yang menghasilkan deret atau urutan angka secara otomatis. Fungsi ini sangat sering digunakan dan dikombinasikan dengan perulangan `for` ketika jumlah iterasi sudah diketahui sebelumnya.

Fungsi `range()` dapat menerima hingga tiga parameter:

- **`start` (opsional)**: Nilai awal dari deret (default bernilai 0 jika tidak dituliskan).
- **`stop`**: Batas akhir deret (angka ini *tidak* termasuk dalam hasil iterasi).
- **`step` (opsional)**: Interval atau selisih antar angka (default bernilai 1).

Contoh penggunaan umumnya adalah `range(1, 10, 2)` yang akan menghasilkan urutan angka: 1, 3, 5, 7, 9.

**2.1.3 Perulangan Bersarang (*Nested Loop*)**

Perulangan bersarang (*nested loop*) adalah kondisi di mana sebuah struktur perulangan berada di dalam tubuh perulangan lainnya. Struktur ini terdiri dari **perulangan luar** (*outer loop*) dan **perulangan dalam** (*inner loop*).

Mekanisme kerjanya adalah:

1. Perulangan luar melakukan satu iterasi.
2. Setiap kali perulangan luar berjalan satu kali, perulangan dalam akan mengeksekusi seluruh siklusnya dari awal hingga selesai.
3. Proses ini berlanjut hingga perulangan luar menyelesaikan seluruh itensitasnya.

Struktur ini sangat berguna dalam pengolahan data berdimensi dua, seperti matriks, pembuatan pola atau bentuk geometris (*pattern printing*), serta tabel perkalian.

**2.1.4 Pengendalian Tampilan Konsol**

Pengendalian tampilan konsol berkaitan dengan bagaimana data atau teks ditampilkan ke layar terminal agar mudah dibaca, terstruktur, dan sesuai dengan format yang diinginkan.

Dalam Python, fungsi `print()` memiliki parameter khusus untuk mengendalikan keluaran (*output*) pada konsol:

- **Parameter `end`**: Mengatur karakter apa yang dicetak di akhir fungsi `print()`. Secara *default*, `end='\n'` (pindah baris baru). Parameter ini dapat diubah (misalnya `end=' '` atau `end=''`) untuk menjaga agar cetakan berikutnya tetap berada di baris yang sama, yang sangat berguna saat membuat pola matriks atau perulangan bersarang.
- **Parameter `sep`**: Mengatur pemisah (*separator*) antar argumen ketika beberapa data dicetak sekaligus dalam satu fungsi `print()` (default berupa spasi).
- **Format Teks**: Penggunaan format string (seperti *f-string*) untuk meratakan teks, mengatur jumlah desimal angka, atau membuat tampilan tabel di konsol menjadi lebih rapi dan simetris.

## 2.2 Kode Program

### 2.2.1 Program "nested_for1_2611532020"

````python
# Buat file dengan nama nested_for1_NIM.py
# Buat program untuk perulangan for dalam Python
# nama variable ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_2020 = int(input("Masukkan nilai batas: "))
for line_2020 in range(1, batas_2020 + 1):
    for j_2020 in range(1, (-1 * line_2020 + batas_2020 + 1)):
        print(".", end="")
    print(line_2020)
````

**Penjelasan:**
1. Baris 1 menggunakan fungsi `input()` untuk meminta nilai batas dari pengguna. Fungsi `int()` mengonversi masukan menjadi bilangan bulat dan menyimpannya pada variabel `batas_2020`.
2. Baris 2 menggunakan fungsi `range()` untuk menjalankan perulangan luar dari 1 sampai `batas_2020`. Nilai batas akhir `batas_2020 + 1` tidak disertakan oleh `range()`, sehingga angka `batas_2020` tetap ikut diproses.
3. Baris 3 menggunakan fungsi `range()` untuk menjalankan perulangan dalam. Jumlah iterasinya berkurang satu pada setiap baris, sehingga jumlah titik yang dicetak juga semakin sedikit.
4. Baris 4 menggunakan fungsi `print(".", end="")` untuk mencetak titik tanpa pindah baris.
5. Baris 5 menggunakan fungsi `print(line_2020)` untuk mencetak angka pada baris tersebut dan berpindah ke baris berikutnya.

### 2.2.4 Program "nested_for4_2611532020"

````python
# Buat file dengan nama nested_for4_NIM.py
# Buat program untuk perulangan for dalam Python
# nama variable ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

tinggi_2020 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_2020 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_2020 = tinggi_2020
    c_2020 = a_2020
    lebar_2020 = (2 * tinggi_2020) - 2
    
    for i_2020 in range(1, tinggi_2020 + 1):
        b_2020 = c_2020 + 1
        
        for j_2020 in range(1, lebar_2020 + 1):
            
            # Baris atas dan bawah
            if i_2020 == 1 or i_2020 == tinggi_2020:
                if j_2020 == 1 or j_2020 == lebar_2020:
                    print("#", end="")
                else:
                    print("=", end="")
            
            # Baris isi
            else:
                if j_2020 == 1 or j_2020 == lebar_2020:
                    print("|", end="")
                else:
                    if j_2020 == c_2020:
                        print("<", end="")
                    elif j_2020 == b_2020:
                        print(">", end="")
                    elif j_2020 == (lebar_2020 - c_2020):
                        print("<", end="")
                    elif j_2020 == (lebar_2020 - c_2020 + 1):
                        print(">", end="")
                    elif j_2020 > b_2020 and j_2020 < (lebar_2020 - c_2020):
                        print(".", end="")
                    else:
                        print(" ", end="")
                        
        print()
        
        # Logika asli Java
        a_2020 -= 1
        
        if a_2020 <= 0:
            c_2020 = (-a_2020) + 2
        else:
            c_2020 = a_2020
````

**Penjelasan:**
1. Baris 1 menggunakan fungsi `input()` untuk meminta tinggi pola. Fungsi `int()` mengonversi masukan menjadi bilangan bulat dan menyimpannya pada variabel `tinggi_2020`.
2. Baris 2 memeriksa apakah `tinggi_2020` merupakan bilangan ganjil. Jika ya, baris 3 menggunakan fungsi `print()` untuk menampilkan pesan bahwa tinggi harus genap.
3. Baris 4 menjalankan cabang `else` jika tinggi bukan bilangan ganjil. Baris 5 sampai 7 menginisialisasi variabel ukuran, posisi, dan lebar pola.
4. Baris 8 menggunakan fungsi `range()` untuk menjalankan perulangan luar dari 1 sampai `tinggi_2020`; setiap iterasi memproses satu baris pola.
5. Baris 9 menentukan posisi simbol pasangan berdasarkan nilai `c_2020`.
6. Baris 10 menggunakan fungsi `range()` untuk menjalankan perulangan dalam dari 1 sampai `lebar_2020`; setiap iterasi memproses satu kolom.
7. Baris 11 sampai 16 membentuk bingkai atas dan bawah. Baris 13 mencetak `#` pada kedua ujung, sedangkan baris 15 mencetak `=` di antaranya.
8. Baris 17 sampai 19 menangani baris isi. Baris 18 mencetak `|` pada kolom tepi pola.
9. Baris 20 sampai 27 memeriksa posisi kolom untuk mencetak pasangan simbol `<` dan `>`.
10. Baris 28 sampai 31 mencetak titik pada bagian pola yang berada di antara simbol, atau spasi pada bagian dalam lainnya.

### 2.2.7 Program "perulangan_for3_2611532020"

````python
# Buat file dengan nama perulangan_for3_NIM.py
# Buat program untuk perulangan for dalam Python
# nama variable ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_2020 = int(input("Masukkan jumlah perulangan: "))

jumlah_2020 = 0
for i in range(1, ulang_2020 + 1):
    print(i, end=" ")
    jumlah_2020 = jumlah_2020 + i

    if i < ulang_2020:
        print(" + ", end=" ")
    else:
        print(" = ", jumlah_2020, end=" ")
print()
print("Jumlah =", jumlah_2020)
````
