## 2.1 Dasar Teori

1. Python merupakan bahasa pemrograman tingkat tinggi yang dirancang dengan filosofi keterbacaan kode melalui sintaks yang bersih, bersifat interpreted, serta dilengkapi pustaka standar yang mendukung berbagai kebutuhan komputasi dari dasar hingga kompleks.
2. Visual Studio Code adalah editor kode sumber ringan yang dilengkapi manajemen ekstensi fleksibel, terminal bawaan, serta lingkungan kerja terintegrasi untuk mendukung proses penulisan dan pengujian kode secara efisien.
3. GitHub adalah platform berbasis cloud untuk pengelolan kode sumber dan penyimpanan repositori menggunakan sistem kontrol versi Git yang memfasilitasi pelacakan perubahan serta kolaborasi proyek secara daring.
4. Visual Studio Code menyediakan integrasi Git bawaan yang memungkinkan pengembang untuk menghubungkan direktori proyek lokal langsung ke repositori GitHub. Melalui fitur Source Control di VS Code, pengguna dapat melakukan inisialisasi repositori, staging, commit perubahan kode, hingga melakukan push dan pull ke GitHub secara langsung tanpa harus keluar dari jendela editor atau menggunakan terminal terpisah.
